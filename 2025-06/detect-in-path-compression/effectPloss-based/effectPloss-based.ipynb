```python
import pandas as pd
import plotly.express as px
df = pd.DataFrame({
    "Country": ["USA", "Germany", "Japan", "Australia"],
    "Loss Rate": [80.5, 78.3, 74.2, 73.1],
    "Error": [5.1, 8.1, 10.2, 12.4]
})
countries = ["USA", "Germany", "Japan", "Australia"]
```

```python
fig = px.bar(
    df,
    x="Country",
    y="Loss Rate",
    error_y="Error",
    labels={"Loss Rate": "Loss Rate"},
    color_discrete_sequence=["green"]
)

# Set width of bars
fig.update_traces(width=0.5)


fig.update_layout(
    width=800, height=540,
    yaxis=dict(range=[0, 102], tick0=0, dtick=20, tickfont = dict(size=14),
               showgrid=True, gridcolor='lightgray', gridwidth=1),
    xaxis=dict(tickfont=dict(size=16), tickangle=45),   # Rotate tick labels by 45 degrees
    yaxis_title=dict(
        text="Loss Rate",
        font=dict(size=20, color="#4D4D4D"),
    ),
    xaxis_title=dict(
        text="",
    ),
    plot_bgcolor="rgba(255,255,255,0)", # transparent
)

fig.add_shape(
    type="rect",
    xref="paper", yref="paper",
    x0=0, y0=0, x1=1, y1=1,
    line=dict(color="#4D4D4D", width=1),
)


fig.show()
```

```python
import plotly.graph_objects as go
```

```python
x_vals = [2, 3, 4, 5]
tick_labels = df["Country"]

fig = go.Figure()

fig.add_trace(go.Bar(
    x=x_vals,
    y=df["Loss Rate"],
    error_y=dict(type="data", array=df["Error"], visible=True),
    width=0.6,
    marker=dict(color='green')
))

fig.update_layout(
    width=800,
    height=600,
    yaxis_title_standoff=13,
    yaxis=dict(
        range=[0, 102],
        tick0=0,
        dtick=20,
        tickfont=dict(size=14, family="Helvetica Neue"),
        ticks='outside',
        tickwidth=1,
        tickcolor='#4D4D4D',
        showline=True,
        linewidth=1.2,
        linecolor='#4D4D4D',
    ),

    xaxis=dict(
        tickmode="array",
        tickvals=x_vals,
        ticktext=tick_labels,
        tickangle=45,
        tickfont=dict(size=16),
        range=[1, 6],
        showline=True,
        linewidth=1.2,
        linecolor='#4D4D4D',
    ),
    yaxis_title=dict(
        text="Loss Rate Difference (%)",
        font=dict(size=20, color="#4D4D4D")
    ),
    xaxis_title=dict(
        text=""
    ),
    plot_bgcolor="rgba(255,255,255,0)"
)

fig.show()
```
