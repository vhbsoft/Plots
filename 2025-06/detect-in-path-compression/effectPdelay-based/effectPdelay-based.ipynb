```python
import pandas as pd
import plotly.express as px
df = pd.DataFrame([
    [10, 0.00825], [20, 0.0165], [50, 0.04125], [100, 0.0825], [200, 0.165], [400, 0.33], [600, 0.495], [800, 0.66], [1000, 0.825]
    ], columns=["Number of Packets", "Delay"])
```

```python
fig = px.line(df, x="Number of Packets", y="Delay", markers=True, line_dash_sequence=["dash"])

fig.update_traces(marker=dict(size=10, symbol="square"), line=dict(width=2.5, color="green"))

fig.update_layout(
    width=800, height=520,

    xaxis=dict(range=[0, 1010], tick0=0, dtick=200, showline=True, ticks='outside', linewidth=1.2, tickwidth=1, linecolor="#4D4D4D",
        tickfont = dict(size=16, family="Helvetica Neue"), tickmode='array',
        tickvals=[i for i in range(0, 1100, 100)],  # ticks every 100
        ticktext=[str(i) if i % 200 == 0 else '' for i in range(0, 1100, 100)]),

    yaxis=dict(range=[0, 1.0], tick0=0, dtick=0.2, showline=True, ticks='outside', linewidth=1.2, tickwidth=1, linecolor="#4D4D4D",
        tickfont = dict(size=16, family="Helvetica Neue"), tickmode='array',
        tickvals=[(i / 1000) for i in range(0, 1100, 100)],  # ticks every 100
        ticktext=['0']+[str(i / 1000) if i % 200 == 0 else '' for i in range(100, 1100, 100)]),

    plot_bgcolor="rgba(255,255,255,0)", # transparent
)

fig.update_layout(
    font=dict(size=16),
    yaxis_title=r"$\Large \Delta t_H - \Delta t_L\,(\mathcal{s})$",
    yaxis_title_standoff=13,
    yaxis_title_font=dict(
        color="#4D4D4D",      # color
    ),
    xaxis_title_font=dict(
        color="#4D4D4D"
    )
)

fig.add_shape(
    type="line",
    xref="paper",  # use full x-axis
    yref="y",      # fixed y-axis
    x0=0,
    x1=1,          # spans right edge
    y0=0.1,
    y1=0.1,
    line=dict(color="black", width=1.5, dash="dot")
)

fig.add_annotation(
    text="100ms Threshold",
    xref="paper", yref="y",
    x=0.79,  # left edge annotation
    y=0.11,
    showarrow=False,
    font=dict(size=14),
    xanchor="left",
    yanchor="bottom"
)

fig.show()
```
