```python
import pandas as pd
import plotly.express as px
df = pd.DataFrame([
    [20, 2.15], [30, 1.3], [40, 0.82], [50, 0.6], [60, 0.41], [70, 0.26], [80, 0.14], [90, 0.07], [100, 0.02]
    ], columns=["Compression Link Capacity (Mbps)", "Delay"])
```

```python
fig_v1 = px.line(df, x="Compression Link Capacity (Mbps)", y="Delay", markers=True, line_dash_sequence=["dash"])

fig_v1.update_traces(marker=dict(size=10, symbol="square"), line=dict(width=2.5, color="green"))

fig_v1.update_layout(
    width=800, height=520,
    xaxis=dict(range=[15, 105], tick0=20, dtick=10, showline=True, ticks='outside', linewidth=1.2, tickwidth=1, linecolor="#4D4D4D",
        tickfont = dict(family="Helvetica Neue", size=16), tickmode='array',
        tickvals=[i for i in range(0, 105, 10)],  # ticks every 10
        ticktext=[str(i) if i % 20 == 0 else '' for i in range(0, 105, 10)]),
    yaxis=dict(range=[0, 2.30], tick0=0, dtick=0.25, showline=True, ticks='outside', linewidth=1.2, tickwidth=1, linecolor="#4D4D4D",
        tickfont = dict(family="Helvetica Neue", size=16), tickmode='array',
        tickvals=[0, 0.25, 0.5, 0.75, 1, 1.25, 1.5, 1.75, 2, 2.25],  # ticks every 0.5
        ticktext=[str(i) if i % 0.5 == 0 else '' for i in [0, 0.25, 0.5, 0.75, 1, 1.25, 1.5, 1.75, 2, 2.25]]),
    plot_bgcolor="rgba(255,255,255,0)", # transparent
)

fig_v1.update_layout(
    font=dict(size=16),
    yaxis_title_standoff=13,
    yaxis_title=r"$\Large \Delta t_H - \Delta t_L\,(\mathcal{s})$",
    yaxis_title_font=dict(
        color="#4D4D4D",      # color
    ),
    xaxis_title_font=dict(
        color="#4D4D4D"
    )
)

fig_v1.add_shape(
    type="line",
    xref="paper",  # use full x-axis
    yref="y",      # fixed y-axis
    x0=0,
    x1=1,          # spans right edge
    y0=0.1,
    y1=0.1,
    line=dict(color="black", width=1.5, dash="dot")
)

fig_v1.add_annotation(
    text="100ms Threshold",
    xref="paper", yref="y",
    x=0.79,  # left edge annotation
    y=0.13,
    showarrow=False,
    font=dict(size=14),
    xanchor="left",
    yanchor="bottom"
)

fig_v1.show()
```
